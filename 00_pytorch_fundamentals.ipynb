###  PyTorch & NumPy

```python
# NumPy array to Tensor
import torch
import numpy as np

array = np.arange(1.0, 8.0)
tensor = torch.from_numpy(array) #اخذت كمان الdefault data type تبعت ال numpy وهو ال float64 الى اذا حددنا ال data type 

array , tensor
```

```python
import numpy as np
print(np.__version__)

import torch
print(torch.__version__)
```

```python
array[1] = 9
array
```

```python
tensor  #same memory
```

```python
array = array + 1
array
```

```python
tensor  
#هل العملية تعدّل الـ array نفسها أم تنشئ واحدة جديدة؟
#فـ torch.from_numpy() يخلي الاثنين يشتركوا في نفس الـ memory، لكن إذا العملية أنتجت object جديد، الـ Tensor القديم ما إله علاقة بالـ object الجديد.
```

```python
# Tensor to NumPy array
tensor = torch.ones(7)
numpy_tensor = tensor.numpy()

tensor , numpy_tensor  # the array takes the default tensor data type, which is float32
```

```python
tensor[0] = 2  # in-place change: numpy_tensor changes too (shared memory)
tensor , numpy_tensor
```

```python
tensor = tensor + 1  # creates a NEW tensor, numpy_tensor no longer follows it
tensor , numpy_tensor
```

### Taking the random out of random

```python
random_tensor_A = torch.rand(3, 4)
random_tensor_B = torch.rand(3, 4)

print(random_tensor_A)
print(random_tensor_B)
print(random_tensor_A == random_tensor_B)
```

```python
RANDOM_SEED = 42

torch.manual_seed(RANDOM_SEED)
random_tensor_C = torch.rand(3, 4)

torch.manual_seed(RANDOM_SEED)
random_tensor_D = torch.rand(3, 4)

print(random_tensor_C)
print(random_tensor_D)
print(random_tensor_C == random_tensor_D)
```
